```python
import pandas as pd
import numpy as np

print("--- STEP 1: LOADING & ENGINEREING FEATURES ---")

# 1. Load the dataset files
X_train = pd.read_csv('X_train.csv')
y_train = pd.read_csv('y_train.csv').values.ravel()
X_test = pd.read_csv('X_test.csv')
y_test = pd.read_csv('y_test.csv').values.ravel()

# 2. Align shapes to prevent any row-mismatch errors
min_train_len = min(len(X_train), len(y_train))
X_tr = X_train.iloc[:min_train_len].copy()
y_tr = y_train[:min_train_len]

min_test_len = min(len(X_test), len(y_test))
X_te = X_test.iloc[:min_test_len].copy()
y_te = y_test[:min_test_len]

# 3. Add Advanced Feature Engineering (Creating Smart Ratios)
# This shows your professor you didn't just dump raw data, but engineered new insights!
for df in [X_tr, X_te]:
    # Check and create packet/byte ratio features safely
    if 'src_bytes' in df.columns and 'dst_bytes' in df.columns:
        df['byte_ratio'] = df['src_bytes'] / (df['dst_bytes'] + 1) # +1 avoids division by zero

    # Fill any potential missing values resulting from division
    df.fillna(0, inplace=True)

print(f"✅ Success! Enhanced X_tr shape: {X_tr.shape}")
print(f"✅ Success! Enhanced X_te shape: {X_te.shape}")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.preprocessing import LabelEncoder

print("--- STEP 1: LOADING & FEATURE ENGINEERING ---")

# 1. Load the dataset files
X_train = pd.read_csv('X_train.csv')
y_train = pd.read_csv('y_train.csv').values.ravel()
X_test = pd.read_csv('X_test.csv')
y_test = pd.read_csv('y_test.csv').values.ravel()

# 2. Align shapes to prevent any row-mismatch errors
min_train_len = min(len(X_train), len(y_train))
X_tr = X_train.iloc[:min_train_len].copy()
y_tr = y_train[:min_train_len]

min_test_len = min(len(X_test), len(y_test))
X_te = X_test.iloc[:min_test_len].copy()
y_te = y_test[:min_test_len]

# 3. Add Advanced Feature Engineering (Creating Smart Ratios)
for df in [X_tr, X_te]:
    if 'src_bytes' in df.columns and 'dst_bytes' in df.columns:
        df['byte_ratio'] = df['src_bytes'] / (df['dst_bytes'] + 1)
    df.fillna(0, inplace=True)

print(f"✅ Data loaded & engineered! X_tr shape: {X_tr.shape}")

print("\n--- STEP 2: TRAINING ADVANCED XGBOOST MODEL ---")

# XGBoost requires target labels to be numeric values
le = LabelEncoder()
y_tr_encoded = le.fit_transform(y_tr)
y_te_encoded = le.transform(y_te)

# Initialize and train the modern XGBoost Classifier
xgb_model = xgb.XGBClassifier(n_estimators=100, learning_rate=0.1, random_state=42, eval_metric='logloss')
print("Training XGBoost model (this may take a short moment)...")
xgb_model.fit(X_tr, y_tr_encoded)
print("✅ XGBoost Model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, classification_report
import joblib

print("--- STEP 3: EVALUATING XGBOOST MODEL ---")

# Make predictions on the test set
y_pred_encoded = xgb_model.predict(X_te)
y_pred = le.inverse_transform(y_pred_encoded)

# Print professional performance metrics
print(f"Accuracy Score: {accuracy_score(y_te, y_pred):.4f}\n")
print("Detailed Classification Report:")
print(classification_report(y_te, y_pred))

print("\n--- STEP 4: SAVING THE MODEL FOR MEMBER 3 ---")

# Save the trained XGBoost model
joblib.dump(xgb_model, 'intrusion_detection_model.pkl')
print("✅ Success! Model saved as 'intrusion_detection_model.pkl'")
```
